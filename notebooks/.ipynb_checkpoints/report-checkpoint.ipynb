## I. Introduction

Solar flares, caused by the tangling of the sun's magnetic field, eject large amounts of electromagnetic energy into space. Although Earth's atmosphere protects us from the majority of emitted charged particles and X-rays, solar flares can disrupt the ionosphere, impacting radio wave propagation around the world. This can be problematic for satellite and air traffic communications. We will analyze a report, "Super-Intense Geomagnetic Storm on 10–11 May 2024: Possible Mechanisms and Impacts" and attempt to recreate their charts based on the open source data provided. This project allows us to test our analystic ability and create our own version and interpretation of a major space weather event in the form of a formal report.

## II. Context

### What causes solar flares?

Solar flares originate from the Sun’s differential rotation—its equatorial and polar areas rotate at different rates—which causes the Sun's magnetic fields to tangle [1]. At these tangled spots, the localized magnetic field can become so strong that it twists away from the Sun's surface, ejecting large amounts of radiation [1]. These ejections create solar storms [1]. Solar storms come in a couple of different flavors [2]. Solar flares are an intense burst of electromagnetic radiation. These range from A (weakest), B, C, M, to X (strongest). These classes are on a logarithmic scale, and within each class, they are further subdivided into 1–9. The energy from flares travels at the speed of light [2]. Radiation storms are explosions of electrically charged particles (protons and electrons) [2]. Coronal mass ejections (CMEs) occur when plasma (electrically charged gas) erupts from the Sun. These are much slower than solar flares and radiation storms but have a much larger impact on Earth. They can induce electrical currents, damage electrical equipment, and heat up the atmosphere, increasing drag on orbiting satellites [2]. Major solar flares can prove a risk to high-altitude flights due to exposure to high-energy particles [2].

## III. Data

Data used in this report:
- [DSCOVR solar wind and IMF](https://www.ngdc.noaa.gov/dscovr/portal/index.html#/)
- [Magnetic field (Bz and By) data from the GOES-18](https://cdaweb.gsfc.nasa.gov/pub/data/goes/goes18/)
- [TEC map data](http://millstonehill.haystack.mit.edu/listExperiments?isGlobal=on%26categories=17%26instruments=8000%26showDefault=on%26start_date_0=1950-01-01%26start_date_1=00%3A00%3A00%26end_date_0=2024-12-31%26end_date_1=23%3A59%3A59)
- [Data from magnetometers, VLF receivers, and Ionosonde](https://zenodo.org/records/12623351)

This data encompassed a broad range of information representative of the impacts of the solar storm recorded from both instruments in space and Earth. The Deep Space Climate Observatory (DSCOVR), a NASA satellite at Lagrange point 1 (L1), recorded solar wind and IMF data using its magnetometer and Faraday cup. The Bz and By magnetic field data was collected from NOAA's Geostationary Operational Environmental Satellites (GOES-18). MIT's Millstone Hill Geospace Facility provided the TEC map data from the worldwide GNSS receiver network. Finally, the magnetometers, VLF receivers and Ionosonde data was available through the Indian Institute of Geomagnetism.

## IV. Methods

Figure 1: GOES and DSCOVR Magnetic Field Components
We accessed magnetic field data from NOAA GOES-18 and NASA DSCOVR via CDAWeb, processing the raw .nc files into Pandas DataFrames. After filtering by the desired time window (20:25–20:40 UT), we plotted Bx, By, and Bz components for both spacecraft to showcase the near-Earth fluctuations (GOES) versus upstream solar wind conditions (DSCOVR), demonstrating the storm’s immediate magnetic impact versus what it was like after closing in on Earth.

Figure 2: DSCOVR Space Weather Parameters
Using the DSCOVR data, we plotted four subpanels: three magnetic components (Bx, By, Bz) and magnitude (Bt), followed by proton speed, density, and temperature. This required parsing multiple variables from DSCOVR .nc files, aligning them temporally, and highlighting the clear solar wind shock arrival around 16:00 UT with Matplotlib.

Figure 3: Solar-Ionospheric Coupling During Flare
GOES X-ray flux data was layered with VLF signal and EEJ values from Indian geomagnetic datasets. We built a three-panel subplot to show solar flare onset and its ionospheric impacts, overlaying May 10 and May 9 for comparison. Ionograms were added by compiling image data from Ionosonde scatterplots at 10-minute intervals, presented in a grid to capture ionospheric restructuring.

Figure 4: TEC Maps Across May 9–11
TEC data from MIT’s Millstone Hill portal were downloaded as global maps and georeferenced onto Plate Carrée projections using Cartopy. After organizing files by timestamp (18:00 and 20:00 UT), we plotted spatial variations using consistent color normalization to reveal enhanced electron content and radio wave interference during the May 10 flare peak.

## V. Results

To understand how the solar flare on May 10, 2024, influenced near-Earth space, we begin by examining the magnetic field conditions measured by spacecraft positioned upstream in the solar wind and within Earth’s magnetosphere. The figure below shows the Bx, By, and Bz components of the magnetic field from two satellite missions: GOES, located in geostationary orbit near Earth, and DSCOVR, positioned at the L1 Lagrange point between Earth and the Sun. These observations help detect changes in the orientation and strength of the interplanetary magnetic field (IMF), which plays a key role in how solar energy enters Earth's magnetosphere. Sudden fluctuations—especially in the Bz component—can signal the arrival of solar wind disturbances or magnetic reconnection events, which often precede geomagnetic storms.



**Figure 1**: Magnetic field components Bx, By, and Bz measured by GOES (gray) and DSCOVR (red) from 20:25 to 20:40 UT on May 10, 2024. GOES data reveal sharp, short-term fluctuations across all three components, consistent with dynamic magnetospheric activity. In contrast, DSCOVR data show smoother upstream solar wind conditions. Together, these measurements provide context for energy transfer into the near-Earth environment during the period of solar activity.

Further analysis on the DSCOVR measurements reveals the intense variations in the magnetic field and a sudden increase in proton speed, density, and temperature upon the arrival of the solar weather on May 10th. DSCOVR, while located at the L1 legrange point, is able to recieve space weather information up to an hour prior to it reaching Earth. 
**Figure 2**: (a) Magnetic field measurements at the DSCOVR sattelite in the x, y, and z directions and the magnetude (bt). (b-d) Proton speed (km/s), proton density (cm^-3), and temperature (deg K) measurements throughout May 10th. Solar activity reached DSCOVR around 16:00 UT.

We can now turn our attention to the earlier stages of the event to better understand how the solar flare influenced Earth's upper atmosphere. The figure below provides a detailed view of the ionospheric response during and shortly after the flare’s peak, beginning around 06:30 UT. Panel (a) shows the sharp increase in X-ray flux detected by satellites, marking the onset of solar-driven energy input. Panels (b) and (c) show ground-based observations of Very Low Frequency (VLF) radio signal amplitude and the strength of the equatorial electrojet (EEJ), both of which respond sensitively to changes in ionospheric ionization. Compared to the quiet reference day (May 9), the data from May 10 show clear evidence of enhanced D-region ionization and current system intensification. On the right, ionograms taken at 10-minute intervals capture how radio wave reflections changed in frequency and altitude, illustrating the short-term restructuring of the ionosphere’s vertical profile in response to the flare. Together, these datasets highlight how solar energy can rapidly couple into and reorganize Earth’s near-space environment.



**Figure 3**: Composite overview of solar and ionospheric conditions on May 10, 2024. (a) X-ray flux from the GOES satellite showing a strong solar flare event. (b) VLF signal amplitude and (c) Equatorial Electrojet (EEJ) strength comparing 10 May (blue) and 09 May (black). Panels (d)–(i) show ionogram data at 10-minute intervals from 0630 UT to 0720 UT on 10 May, with altitude vs. frequency scatter plots color-coded by signal-to-noise ratio (SNR). The ionograms reveal dynamic changes in ionospheric reflectivity associated with the flare event.

Lastly, we analyzized the total electron content (TEC) during May 9th, 10th, and 11th at two separate time stamps (18:00 and 20:00 UT). 

TEC, measured in electrons per square meter, has the ability to impeede radio wave transmissions. It is calculated through a dual frequency reciever by measuring the delay between frquencies. More electrons along the path of a radio wave between its transmitter and reciever will have a greater overall impact on the radio's signal. Sometimes, with especially high TEC values, radio waves can be refracted and scattered leading to radio blackouts. Additionally, high TEC values can disrupt global possitioning systems like GPS.

On May 9th, the TEC concentration remained relatively constant across both time stamps. Elevated TEC activity can be seen in North and South America, while relatively low levels are found accross other parts of the world. On May 10th, the TEC becomes elevated from 18:00 to 20:00. These higher levels are induced by the arrival of the solar storm. By the 11th at 18:00 the TEC values across the globe are relatively low.



**Figure 4:** TEC data overlain on plate carrée map projections at 18:00 and 20:00 timestamps between May 9th and May 11th.

## VI. Challenges

In our attempts to recreate many of the graphs present in the "Super-Intense Geomagnetic Storm on 10–11 May 2024: Possible Mechanisms and Impacts" report, we encountered many complications. Some of the cited "Open Research" sources were wholly inaccessible or very difficult to locate. For example, in just Figure 1 and Figure 2, there were three different file formats, ".nc", ".txt", and ".csv". Even after opening the data and transforming them into datasets or dataframe's issues remained, whether it be with excessive missing values, incorrectly parsed files, or variations in column and feature names.

## VII. Conclusion

Overall, the May 2024 solar storm was a significant event. The high magnitude strength solar flares and coronal mass ejections during this event had major effects on the conditions of the ionosphere, causing peaks in the VLF signal amplitude and elevated TEC values. Auroras, communication interferences, and GPS inaccuracies were among some of the impacts. By analyzing these events, we can better understand and prepare for future solar weather.
