## I. Introduction

Solar flares, caused by the tangling of the sun's magnetic field, eject large amounts of electromagnetic energy into space. Although Earth's atmosphere protects us from the majority of emitted charged particles and X-rays, solar flares can disrupt the ionosphere, impacting radio wave propagation around the world. This can be problematic for satellite and air traffic communications. We will analyze a report, ["Super-Intense Geomagnetic Storm on 10–11 May 2024: Possible Mechanisms and Impacts,"](https://agupubs.onlinelibrary.wiley.com/doi/full/10.1029/2024SW004126#swe21812-bib-0020) and attempt to recreate their charts based on the open-source data provided. This project allows us to test our analytical ability and create our own version and interpretation of a major space weather event in the form of a formal report.

### May 10-11 2024 Geomagnetic Storm

***"This is an unusual and potentially historic event."*** 

     - Clinton Wallace, Director, NOAA’s Space Weather Prediction Center [4]

- **The biggest solar storm to hit Earth in 20 years**
- The Sun neared its peak solar activity period (solar maximum) in its 11-year solar cycle
- Released several significant solar flares, stretching our protective atmosphere to its limits

## II. Data and Methods
- [DSCOVR Space Weather Data](https://www.ngdc.noaa.gov/dscovr/portal/index.html#/) - Magnetic field and proton measurements
- [GOES-18 Satellite Data](https://cdaweb.gsfc.nasa.gov/pub/data/goes/goes18/) - Magnetic field
- [Data from magnetometers, VLF receivers, and Ionosonde](https://zenodo.org/records/12623351)
- [TEC map data](http://millstonehill.haystack.mit.edu/)

This data encompassed a broad range of information representative of the impacts of the solar storm recorded from both instruments in space and on Earth. The Deep Space Climate Observatory (DSCOVR), a NASA satellite at Lagrange point 1 (L1), recorded solar wind and IMF data using its magnetometer and Faraday cup. The Bz and By magnetic field data were collected from NOAA's Geostationary Operational Environmental Satellites (GOES-18). MIT's Millstone Hill Geospace Facility provided the TEC map data from the worldwide GNSS receiver network. Finally, the magnetometers, VLF receivers, and Ionosonde data were available through the Indian Institute of Geomagnetism.

To analyze the data above, we used the following packages:
- xarray
- numpy
- matplotlib
- pandas
- cartopy

Our analysis is split into two main sections. First, we discuss satellite magnetic field measurements, which provide indications of the arrival of the solar weather and context to the overall magnitude of the event. Then, through analysis of very low-frequency amplitudes and total electron content, we look at the impact of the solar storm on the ionosphere and, subsequently, radio signals.

## III. Results

### i. Satellite Measurements



**Figure 1**: The magnetic field vectors of Bx, By, and Bz as measured by GOES (gray) and DSCOVR (red) satellites from 20:25 to 20:40 UT on May 10, 2024. These measurements show fluctuations in the magnetic field's strength in a specific direction (along the x, y, or z axis). Large variations and values, especially in Bz (perpendicular to our ecliptic), are caused by strong waves and disturbances in the solar wind [7].



**Figure 2**: (a) Magnetic field measurements at the DSCOVR satellite in the Bx, By, and Bz directions and the magnitude (Bt). (b)-(d) Proton speed (km/s), proton density (cm^-3), and temperature (deg K) measurements throughout May 10th. Solar activity reached DSCOVR around 16:00 UT when intense variations in magnetic field components and sudden increases in proton measurements occurred.

**Figure 3:** (a) Magnetic field measurements at the DSCOVR satellite in the Bx, By, and Bz directions and the magnitude (Bt). (b)-(d) Proton speed (km/s), proton density (cm^-3), and temperature (deg K) measurements throughout May 11th. The solar activity begins to wind down at the DSCOVR satellite around 12:00 UT. The proton measurements experience a quicker change back to normal, while the magnetic field changes are more gradual.

### ii. Ionosphere Impacts



**Figure 4**: Overview of both the solar and ionospheric conditions during the May 10, 2024, solar storm. X-ray flux is a measure of solar activity, VLF (very low frequency) signal amplitude is a representation of wave activity, and EEJ (Equatorial Electrojet) is the measurement of the electric current that flows in the equatorial region of Earth's ionosphere. The scatterplots on the right side of the compiled figure show ionogram (ions) data plotted by altitude vs. frequency and color coded based on the signal-to-noise ratio (the higher SNR values indicate a surplus of noise compared to comprehensible data at higher elevations and frequencies).



**Figure 5:** TEC data overlain on plate carrée map projections at 18:00 and 20:00 UT timestamps for May 9th through May 11th. On May 9th, the TEC concentration remained relatively constant across both time stamps. Elevated TEC activity can be seen in North and South America, while relatively low levels are found across other parts of the world. On May 10th, the TEC becomes elevated from 18:00 to 20:00. These higher levels are induced by the arrival of the solar storm. By the 11th at 18:00, the TEC values across the globe are relatively low.

## IV. Challenges

In our attempts to recreate many of the graphs present in the "Super-Intense Geomagnetic Storm on 10–11 May 2024: Possible Mechanisms and Impacts" report, we encountered many complications. Some of the cited "Open Research" sources were wholly inaccessible or very difficult to locate. Additionally, the few datasets we were able to acquire were organized in different ways and data formats. For example, in just Figure 1 and Figure 2, there were three different file formats, ".nc", ".txt", and ".csv". Even after opening the data and transforming them into datasets or dataframe's issues remained, whether it be with excessive missing values, incorrectly parsed files, or variations in column and feature names.

## VI. Conclusion

Overall, the May 2024 solar storm was a significant event. The high magnitude strength solar flares and coronal mass ejections during this event had major effects on the conditions of the ionosphere, causing peaks in the VLF signal amplitude and elevated TEC values. Auroras, communication interferences, and GPS inaccuracies were among some of the impacts. By analyzing these events, we can better understand and prepare for future solar weather.
