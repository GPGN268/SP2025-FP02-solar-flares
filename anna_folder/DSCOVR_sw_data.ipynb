# DSCOVR Space Weather Data

The data explored in this notebook contains magnetic field and proton data from the Deep Space Climate Observatory (DSCOVR) Satelite. NOAA deployed DSCOVR in 2015 to the L1 Legrange point in the Sun-Earth system. The L1 point is a neutral gravitational area, allowing the DSCOVR satellite to keep a constant orbiting position between the Earth and the Sun- optimal for receiving solar weather information up to an hour before it reaches Earth. This helps to provide early information about the strength of solar weather prior to it hitting Earth. Learn more about DSCOVR [here](https://www.nesdis.noaa.gov/current-satellite-missions/currently-flying/dscovr-deep-space-climate-observatory). Learn more about Legrangian points [here](https://science.nasa.gov/resource/what-is-a-lagrange-point/).

Aboard DSCOVR, the magnetometer and Faraday cup instruments measure information on space weather including the magnetic field vector and various proton attributes. [Data](https://www.ngdc.noaa.gov/dscovr/portal/index.html#/download/1715126400000;1715472000000/f1m;m1m) from DISCOVR is in a NetCDF format, so the xarray, numpy, and matplotlib python tools were used in this analysis. Each file contains many different features regarding the measurements and specific flaggings specified by NOAA. We selected for just magnetic field measurements in three dimensions, the magnitude of the magnetic field, proton speed, proton density, and proton temperature.

```python
import numpy as np
import xarray as xr
from matplotlib import pyplot as plt
```

## May 8th, 2024

```python
# load data from DSCOVR Magnetometer Level 2 One Minute Averages
ds_raw = xr.load_dataset('../data/DSCOVR_data/may_8/oe_m1m_dscovr_s20240508000000_e20240508235959_p20240509021546_pub.nc')
magnetometer_variables = ['bx_gsm', 'by_gsm', 'bz_gsm', 'bt']
mag_ds8 = ds_raw[magnetometer_variables]

# load data from DSCOVR Faraday Cup Level 2 One Minute Averages
ds_raw = xr.load_dataset('../data/DSCOVR_data/may_8/oe_f1m_dscovr_s20240508000000_e20240508235959_p20240509022123_pub.nc')
farraday_variables = ['proton_speed', 'proton_density', 'proton_temperature']
far_ds8 = ds_raw[farraday_variables]
```

```python
fig, ax = plt.subplots(4, 1, figsize=(20,10), sharex='col', facecolor='grey')
ax1, ax2, ax3, ax4 = ax

### Magnetometer data
# time series plot for bx_gsm
ax1.plot(mag_ds8.time, mag_ds8.bx_gsm, linewidth=1, color='b', label='bx gsm')
ax1.set_title('DSCOVR May 8th, 2024 Space Weather Measurements', fontsize=15)
ax1.set_ylabel('Mag (nT)')
# time series plot for by_gsm
ax1.plot(mag_ds8.time, mag_ds8.by_gsm, linewidth=1, color='g', label='by gsm')
# time series plot for bz_gsm
ax1.plot(mag_ds8.time, mag_ds8.bz_gsm, linewidth=1, color='r', label='bz gsm')
# time series plot for bt
ax1.plot(mag_ds8.time, mag_ds8.bt, linewidth=1, color='k', label='bt')
ax1.set_ylim(-15, 15)
#plt.xlim(05-08 00, 05-09-00)
ax1.legend(loc='upper right', ncol=4)
ax1.grid()
ax1.set_facecolor('#dce0e7cc')

### Farraday data
# time series plot for proton speed
ax2.plot(far_ds8.time, far_ds8.proton_speed, linewidth=1, color='purple', label='proton speed')
ax2.set_ylabel('Speed [Km/s]')
ax2.legend()
ax2.grid()
ax2.set_facecolor('#dce0e7cc')
# time series plot for proton_density
ax3.plot(far_ds8.time, far_ds8.proton_density, linewidth=1, color='teal', label='proton density')
ax3.set_ylabel('Density [1/cm^3]')
ax3.legend()
ax3.grid()
ax3.set_facecolor('#dce0e7cc')
# time series plot for proton_temperature
ax4.plot(far_ds8.time, far_ds8.proton_temperature, linewidth=1, color='brown', label='proton temperature')
ax4.set_ylabel('Temperature [degrees K]')
ax4.ticklabel_format(axis='y', style='sci', scilimits=(0, 0))
ax4.legend()
ax4.grid()
ax4.set_facecolor('#dce0e7cc')

plt.show()
```
