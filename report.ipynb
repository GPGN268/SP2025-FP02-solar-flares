## I. Introduction

Solar flares, caused by the tangling of the sun's magnetic field, eject large amounts of electromagnetic energy into space. Although Earth's atmosphere protects us from the majority of emitted charged particles and X-rays, solar flares can disrupt the ionosphere, impacting radio wave propagation around the world. This can be problematic for satellite and air traffic communications. ## II. Context

### What causes solar flares?
- The equatorial and polar areas of the Sun rotate at different rates, which causes the Sun's magnetic fields to tangle. [1]
- At these tangled spots, the localized magnetic field can become so strong that they twist away from the Sun's surface, ejecting large amounts of radiation. [1]
- These ejections create solar storms. [1]
- Solar storms come in a couple of different flavors: [2]
    - Solar flares, which are an intense burst of electromagnetic radiation. These range from A (weakest), B, C, M, X(strongest). These classes are on a logarithmic scale. Within each class, they are further subdivided into 1-9.
        - Energy from flares travel at the speed of light!
    - Radiation storms are explosions of electrically charged particals (protons and electrons).
    - Coronal mass ejections (CMEs) occur when plasma (electrically charged gas) erupts from the sun. These are much slower than solar flares and radiation storms but have a much larger impact on Earth. They can induce electrical currents and damage electrical equipment and heat up the atmosphere, increasing drag on orbiting satellites.
- Major solar flares can prove a risk to high altitude flights due to exposure to high-energy particles.

Data used in this report:

- [Data from magnetometers, VLF receivers, and Ionosonde](https://zenodo.org/records/12623351)

This data encompassed a broad range of information representative of the impacts of the solar storm recorded from both instruments in space and Earth. The Deep Space Climate Observatory (DSCOVR), a NASA satellite at Lagrange point 1 (L1), recorded solar wind and IMF data using its magnetometer and Faraday cup. The Bz and By magnetic field data was collected from NOAA's Geostationary Operational Environmental Satellites (GOES-18). MIT's Millstone Hill Geospace Facility provided the TEC map data from the worldwide GNSS receiver network. Finally, the magnetometers, VLF receivers and Ionosonde data was available through the Indian Institute of Geomagnetism.

## IV. Results

To understand how the solar flare on May 10, 2024, influenced near-Earth space, we begin by examining the magnetic field conditions measured by spacecraft positioned upstream in the solar wind and within Earth’s magnetosphere. The figure below shows the Bx, By, and Bz components of the magnetic field from two satellite missions: GOES, located in geostationary orbit near Earth, and DSCOVR, positioned at the L1 Lagrange point between Earth and the Sun. These observations help detect changes in the orientation and strength of the interplanetary magnetic field (IMF), which plays a key role in how solar energy enters Earth's magnetosphere. Sudden fluctuations—especially in the Bz component—can signal the arrival of solar wind disturbances or magnetic reconnection events, which often precede geomagnetic storms.



**Figure 1**: Magnetic field components Bx, By, and Bz measured by GOES (gray) and DSCOVR (red) from 20:25 to 20:40 UT on May 10, 2024. GOES data reveal sharp, short-term fluctuations across all three components, consistent with dynamic magnetospheric activity. In contrast, DSCOVR data show smoother upstream solar wind conditions. Together, these measurements provide context for energy transfer into the near-Earth environment during the period of solar activity.

We can now turn our attention to the earlier stages of the event to better understand how the solar flare influenced Earth's upper atmosphere. The figure below provides a detailed view of the ionospheric response during and shortly after the flare’s peak, beginning around 06:30 UT. Panel (a) shows the sharp increase in X-ray flux detected by satellites, marking the onset of solar-driven energy input. Panels (b) and (c) show ground-based observations of Very Low Frequency (VLF) radio signal amplitude and the strength of the equatorial electrojet (EEJ), both of which respond sensitively to changes in ionospheric ionization. Compared to the quiet reference day (May 9), the data from May 10 show clear evidence of enhanced D-region ionization and current system intensification. On the right, ionograms taken at 10-minute intervals capture how radio wave reflections changed in frequency and altitude, illustrating the short-term restructuring of the ionosphere’s vertical profile in response to the flare. Together, these datasets highlight how solar energy can rapidly couple into and reorganize Earth’s near-space environment.



**Figure 2**: Composite overview of solar and ionospheric conditions on May 10, 2024. (a) X-ray flux from the GOES satellite showing a strong solar flare event. (b) VLF signal amplitude and (c) Equatorial Electrojet (EEJ) strength comparing 10 May (blue) and 09 May (black). Panels (d)–(i) show ionogram data at 10-minute intervals from 0630 UT to 0720 UT, with altitude vs. frequency scatter plots color-coded by signal-to-noise ratio (SNR). The ionograms reveal dynamic changes in ionospheric reflectivity associated with the flare event.
